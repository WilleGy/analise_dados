```python
import pandas as pd
import openpyxl
```

```python
path = r"Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"
```

```python
indicador_2 = pd.read_excel(path, engine="openpyxl")
```

```python
indicador_2.columns = [
    "Região", "Total", "Total_Homens", "Total_Mulheres",
    "Exatas_Total", "Exatas_Homens", "Exatas_Mulheres",
    "Humanas_Total", "Humanas_Homens", "Humanas_Mulheres",
]
```

```python
indicador_2
```

```python
indicador_2 = indicador_2.drop(index=0)
indicador_2 = indicador_2.iloc[:-1]
indicador_2 = indicador_2.reset_index(drop=True)
```

```python
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
regiao = indicador_2[indicador_2["Região"].isin(regioes)].reset_index(drop=True)

colunas = regiao.columns[1:]
regiao[colunas] = regiao[colunas].astype(int)
```

```python
table1 = regiao[["Região", "Total", "Total_Homens", "Total_Mulheres"]].copy()
table1.columns = ["Região", "Total", "Homens", "Mulheres"]
```

```python
table1
```

---------------------------------------------------------------------

```python
table2 = regiao[["Região", "Exatas_Total", "Exatas_Homens", "Exatas_Mulheres"]].copy()
table2.columns = ["Região", "Total", "Homens", "Mulheres"]
```

```python
table2
```

---------------------------------------------------------------------

```python
table3 = regiao[["Região", "Humanas_Total", "Humanas_Homens", "Humanas_Mulheres"]].copy()
table3.columns = ["Região", "Total", "Homens", "Mulheres"]
```

```python
table3
```

---------------------------------------------------------------------

```python
table1.to_excel("tabela1.xlsx", index=False)
table2.to_excel("tabela2.xlsx", index=False)
table3.to_excel("tabela3.xlsx", index=False)
```
