```python
import pandas as pd
import openpyxl

import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
path = r"H:\Python\3Etapa\analise_dados\atividade4\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"
```

```python
dados_bruto = pd.read_excel(path, header=2)
```

```python
dados_bruto
```

```python
dados_trabalhados = dados_bruto.iloc[0:5570, 1:]
```

```python
dados_trabalhados
```

```python
tabela_trabalhada = dados_trabalhados
```

```python
tabela_trabalhada
```

```python
tabela_trabalhada["diferenca"] = tabela_trabalhada["População Censo 2022"] - tabela_trabalhada["População 2010 (Alterações de Limites até 2022)1"]
```

```python
tabela_trabalhada
```

```python
estados = tabela_trabalhada.groupby("UF")[["População 2010 (Alterações de Limites até 2022)1", "População Censo 2022", "diferenca"]].sum().reset_index()
```

```python
estados
```

```python
estados.to_csv(r"H:\Python\3Etapa\analise_dados\atividade4\pop_estado.csv", sep = ";")
```

```python
cidades = tabela_trabalhada.iloc[:, 3:]
```

```python
cidades
```

```python
cidades.to_csv(r"H:\Python\3Etapa\analise_dados\atividade4\pop_municipal.csv", sep=";")
```

```python
fig, ax = plt.subplots(figsize=(9, 8))
ordem = estados.sort_values("População Censo 2022", ascending=False)
sns.barplot(data=ordem, y="UF", x="População Censo 2022", ax=ax, color="steelblue")

media_populacao = ordem["População Censo 2022"].mean()
ax.axvline(
    x=media_populacao,
    color="crimson",
    linestyle="--",
    linewidth=2,
)

ax.set_title("População 2022 por UF")
ax.set_xlabel("Habitantes")
ax.set_ylabel("UF")
plt.tight_layout()
plt.show()
```
